# Unit testing

Unit tests check individual parts of a program in isolation.

A test is a piece of code that checks a specific part of a system.

A test has three main steps:

- Arrange the test data.
- Act by running the code under test.
- Assert the expected result, such as a return value or a database write.

Principles for writing tests:

- Keep tests simple and readable.
- Check one behavior per test.
- Verify the actual result without changing it to match expectations.
- Test what a function should do, rather than how it is implemented.
- Keep tests fast and economical with resources.

## Create a more complex test

````python
# Create a Student class.
class Student:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
    
    # Expose get_full_name as a property.
    # It can be read without calling get_full_name().
    # Access it as an attribute: .get_full_name.
    @property
    def get_full_name(self):
        return '{} {}'.format(self.first_name, self.last_name)
    
    @property
    def get_all_data_on_student(self):
        return 'First name: {}, Last name: {}, age is: {}'.format(self.first_name, self.last_name, self.age)
````

````python
# Write the tests.
import unittest


class TestStudent(unittest.TestCase):
    # setUp runs before each test; create the objects here.
    def setUp(self):
        print('setUp')
        self.student_1 = Student(first_name='Alex', last_name='Girlea', age=30)
        self.student_2 = Student(first_name='Damian', last_name='Folvaiter', age=31)
        
    # tearDown runs after each test.
    def tearDown(self):
        print('tearDown')
    
    def test_get_full_name(self):
        print('test_get_full_name')
        self.assertEqual(self.student_1.get_full_name, 'Alex Girlea')
        self.assertEqual(self.student_2.get_full_name, 'Damian Folvaiter')
    
    def test_get_all_data_on_student(self):
        print('test_get_all_data_on_student')
        self.assertEqual(self.student_1.get_all_data_on_student, 'First name: Alex, Last name: Girlea, age is: 30')
        self.assertEqual(self.student_2.get_all_data_on_student, 'First name: Damian, Last name: Folvaiter, age is: 31')
````

````python
# setUp runs before the test and tearDown runs afterward.
# Use tearDown to clean up resources.

# Test methods do not necessarily run in the order they appear in the source.
unittest.main(argv=['first-arg-is-ignored'], exit=False)
````

````python
# Create a Student class.
class Student:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        if type(age) != int:
            raise TypeError
        self.age = age
    
    # Expose get_full_name as a property.
    # It can be read without calling get_full_name().
    # Access it as an attribute: .get_full_name.
    @property
    def get_full_name(self):
        return '{} {}'.format(self.first_name, self.last_name)
    
    @property
    def get_all_data_on_student(self):
        return 'First name: {}, Last name: {}, age is: {}'.format(self.first_name, self.last_name, self.age)
    
    def check_age_limit(self,n):
        if type(n) != int:
            raise TypeError
        else:
            # Perform the operation here.
            pass
````

````python
import unittest

class TestStudent(unittest.TestCase):
    # setUpClass runs once before this test class, for example to load shared fixtures.
    @classmethod
    def setUpClass(cls):
        print('\n classSetUp')
    
    # tearDownClass runs once after this test class.
    @classmethod
    def tearDownClass(cls):
        print('\n tearDownClass')
    
    # setUp runs before each test.
    def setUp(self):
        print('\n setUp', '-'*10)
        self.student_1 = Student(first_name='Alex', last_name='Girlea', age=30)
        self.student_2 = Student(first_name='Damian', last_name='Folvaiter', age=31)
        
    # tearDown runs after each test.
    def tearDown(self):
        print('\n tearDown', '_'*10)
    
    def test_get_full_name(self):
        print('\n test_get_full_name')
        self.assertEqual(self.student_1.get_full_name, 'Alex Girlea')
        self.assertEqual(self.student_2.get_full_name, 'Damian Folvaiter')
    
    def test_get_all_data_on_student(self):
        print('\n test_get_all_data_on_student')
        self.assertEqual(self.student_1.get_all_data_on_student, 'First name: Alex, Last name: Girlea, age is: 30')
        self.assertEqual(self.student_2.get_all_data_on_student, 'First name: Damian, Last name: Folvaiter, age is: 31')
    
    def test_check_age_limit(self):
        print('\n check_age_limit')
        # Check whether passing a string raises TypeError.
        # The test passes if TypeError is raised.
        self.assertRaises(TypeError, self.student_1.check_age_limit, '123')
                        
````
