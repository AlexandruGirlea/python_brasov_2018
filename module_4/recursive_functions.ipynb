### A function is recursive when it calls itself to produce a result.

```python
def factorial(n):
    if n in (0, 1): # base case
        return 1
    return factorial(n - 1) * n # recursive case

factorial(4) # 1 x 2 x 3 x 4
```

```python
def factorial(n):
    return factorial(n - 1) * n # Recursive case

factorial(4) # 1 x 2 x 3 x 4
# Without a base case, this exceeds the maximum recursion depth.
# Expected error: maximum recursion depth exceeded
```

```python
import sys
sys.setrecursionlimit(10000)
# Increasing the limit is not a substitute for a valid base case.
```
