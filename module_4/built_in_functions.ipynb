## Every function is a fully-fledged object and, as such, they have many attributes.

```python
def multiplication(a, b=2):
    return a * b
```

```python
# The default value of b is 2.
# Inspect the attributes of a function.
multiplication.__defaults__
```

```python
multiplication.__code__
```

```python
special_attributes = dir(multiplication)
special_attributes
```

```python
# Retrieve the attributes.
for attribute in special_attributes:
    print('\n')
    print(attribute, '->', getattr(multiplication, attribute))
```

```python
# List the built-in names.
dir(__builtins__)
```

```python
# built in ZIP function

a = (0,1,2,3,4)
b = (10,11,12,13, 14, 15)

list(zip(a,b))
```

```python
# built in TYPE
type(a)
```

```python
# built in RANGE
c = range(0,10)
c
```
