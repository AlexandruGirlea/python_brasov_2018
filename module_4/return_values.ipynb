### Functions return a single object; a tuple can hold multiple return values.

```python
# A function without an explicit return value returns None.

def func():
    pass

a = func()

print(a)
```

```python
def func(n):
    if n in [1,2,3]:
        return n
    else:
        return 0

func(1)
```

```python
func(5)
```

```python
from functools import reduce
from operator import mul

# Homework: explain how this function works (original assignment: submit by email).
def factorial(n):
    return reduce(mul, range(1, n + 1), 1)

f5 = factorial(5)

f5
```

```python
# A Python function can return the result of another function.
def func(*n):
    return min(n)

func(1,3,4,-7)
    
```

```python
# Return multiple values as a tuple.
def two_value_function():
    return 1,False

a, b = two_value_function()
a
```

# Function design principles

- Give each function one clear responsibility.
- Keep functions short.
- Use as few parameters as practical.
- Think carefully about the return value.
- Avoid unexpected side effects.
