## Anonymous functions are called lambdas in Python.

A lambda expression creates a function without requiring a named function definition.

```python
def func(a, b):
    return(a + b)

a = func(10, 11)
a
```

```python
a = lambda x, y: x + y

a(10, 11)
```

```python
# map takes a function and an iterable.
# It applies the function to each item.

def func(x):
    return x * 2

map(func, [1,2,3])
```

```python
list(map(func, [1,2,3]))
```

```python
# A one-line function used only in this expression
list(map(lambda x : x*2, [1, 2, 3]))
```

```python
list_1 = [1,2,3,10]

newList = filter(lambda x: x / 2 == 5, list_1)
for i in newList:
    print(i)
```

```python
a = lambda x: x + 1
a(1)
```

```python
def func(n):
    return n+1
func(1)
```

```python
a = lambda x: x / 2 == 5
a(10)
```
