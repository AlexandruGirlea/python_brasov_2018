## Why use functions?
Functions are among the most important concepts and constructs of any language,
so let me give you a few reasons why we need them:
- They reduce code duplication in a program. By having a specific task
taken care of by a nice block of packaged code that we can import and
call whenever we want, we don't need to duplicate its implementation.
- They help in splitting a complex task or procedure into smaller blocks,
each of which becomes a function.
- They hide the implementation details from their users.
- They improve traceability.
- They improve readability.

https://docs.python.org/2/library/functions.html

```python
def order(*args, **kwargs):
    if args:
        for i in args:
            print(i)
    if kwargs:
        for i in kwargs:
            print(i)
        print(kwargs)
order('Alex', 'Adi', 'Andrei')

# Dictionary order was not guaranteed in the Python 3.5 course environment.
order(key_1='abcd', key_2='abcd') 
```

# Scopes
Scope types:
- local;
- enclosing;
- global;
- built-in.

```python
test = 0

def function_1():
    test = 1
    print('test equals: ', test)

function_1()
```

```python
test
```

```python
def outer():
    test = 1 # outer scope
    def inner():
        test = 2 # inner scope
        print('inner:', test)
    inner()
    print('outer:', test)
    
test = 0 # global scope
outer()

print('global:', test)
```

# Passing arguments

- Passing an argument binds a local parameter name to an object.

```python
x = 2

def func(y):
    print(y)
func(x)
```

- Reassigning a local parameter does not reassign the name used by the caller.

```python
x = 2
def func(x):
    x = 7

func(x)
print(x)
```

- Functions can modify mutable objects.

```python
a = [1,2,3,4]

def func():
    a[0] = 10

func()
a
```

```python
a = [1,2,3,4]

def func():
    a[0] = 10
    a = 'this is a string'

func()
print(a)
# This raises an error because the assignment makes a local to the entire function.
```

```python
a = [1,2,3,4]

def func(a):
    a[0] = 10
    a = 'this is a string'
    print('variable a inside the function: ', a)

func(a)
print(a)
# The shared list was modified, while the local name a was then rebound to a string.
```

## Types of arguments

- Positional arguments (the most common)
- Keyword arguments and default values
- Variable positional arguments
- Variable keyword arguments
- Keyword-only arguments

1. Positional arguments (the most common)

```python
def func(a, b, c):
    print(a, b, c)

func(1,2,3)
```

2. Keyword arguments and default values

Keyword arguments are passed using the `name=value` syntax.

```python
func(a=1, c=2, b=3)
```

```python
# Use default values.
def func(a, b=10, c=11):
    print(a, b, c)
    
func(1)
func(a=1, b=2, c=3)
```

3. Variable positional arguments: `*args` collects positional arguments into a tuple; `*` can also unpack an iterable in a call.

```python
# Find the smallest value in a tuple.
def minimum(*n):
    # print(n)  # n is a tuple
    if n: # Continue if the tuple is not empty.
        smallest = n[0]
        for value in n[1:]:
            if value < smallest:
                smallest = value
        print(smallest)
            
minimum(1,3,4,5,-1)
```

```python
n = (1,3,4,5,-1)
n
```

```python
n[1:]
```

```python
n[:-1]
```

4. Variable keyword arguments -> syntax ( ** instead of * )

```python
def func(**kwargs):
    print(kwargs)

func()
```

```python
func(a=1, b=2)
```

```python
a = {'a':123}
a.get('b')
```

5. Keyword-only arguments are declared after `*args` or a bare `*`.

```python
def func(*a, b):
    print(a, b)

func(1,2,3, b=4)
```

```python
func(b=4)
```

```python
func(1,2,3,4)
```

```python
def func(*, b):
    print(b)

func(1)
```

```python
func(b=1)
```

# Putting it together

```python
def func(a, b, c=10, *args, **kwargs):
    print('a, b, c:', a, b, c)
    print('args:', args)
    print('kwargs:', kwargs)

func(1, 2, 3, 5, 7, 9, A='a', B='b')
# c is 3 rather than its default value of 10 because it was supplied positionally.
```
