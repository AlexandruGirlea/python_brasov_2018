## Cod scurt repetitiv, pe care nu vrem sa il scriem de fiecare data

```python
# exemplu functie normla

import time

def a():
    time.sleep(0.2)

def b():
    time.sleep(0.3)

initial_time = time.time()

a()

print('timpul de executie pentru functia a, este: ', time.time()-initial_time)


initial_time = time.time()

b()

print('timpul de executie pentru functia b, este: ', time.time()-initial_time)
```

```python
# pentru a nu mai scrie time.time() - time putem face o funtie

def time_measure(func):
    initial_time = time.time()
    func()
    print('Functia', func.__name__, 'dureaza:', time.time() - initial_time)
```

```python
time_measure(a)
time_measure(b)
```

```python
# acuma vedem ce se intampla daca functiile a si b primesc parametrii
def a(sleep_time):
    time.sleep(sleep_time)
    
def b(sleep_time):
    time.sleep(sleep_time)
```

```python
# daca incercam sa trimitem asa time_sleep primim o eroare pentru ca in interiorul lui
# time_measure, func() trebuie sa primeasca parametrii
time_measure(a(1))
```

```python
# definim o functie in alta functie pentru ca avem nevoie de o metoda interna care sa se ocupe
# de trimiterea unor argumente catre functia initiala
def time_measure(func):
    
    def wrapper(*args, **kwargs):
        initial_time = time.time()
        func(*args, **kwargs)
        print('Functia', func.__name__, 'dureaza:', time.time() - initial_time)
    # return-ul de fapt cheama functia de mai sus, wrapper primeste parametrii args si kwargs
    # si ii trimite mai jos catre functie
    return wrapper
```

```python
a = time_measure(a)
a(sleep_time=1)
```

```python
# pentru a pastra atributele functiei initiale cand folosim un decorator, vom folosi o librarie
from functools import wraps

def time_measure(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        initial_time = time.time()
        func(*args, **kwargs)
        print('Functia', func.__name__, 'dureaza:', time.time() - initial_time)
    # return-ul de fapt cheama functia de mai sus, wrapper primeste parametrii args si kwargs
    # si ii trimite mai jos catre functie
    return wrapper
```

```python
@time_measure
def a(sleep_time):
    time.sleep(sleep_time)

a(1)
```

```python
# dupa cum putem vedea functia "a" si-a pastrat toate atributele
a.__name__
```

```python
# un exemplu practic de decorator
def minimum_income(income):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = func(*args, **kwargs)
            if result > income:
                print('venitul este ok')
            else:
                print('venitul este prea mic')
        return wrapper
    return decorator

@minimum_income(2000)
def income_calculator(x):
    return x * 2

income_calculator(100)
```
