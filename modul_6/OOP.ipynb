# Attribute shadowing

````python
# prin atribute shadowing ne referim la urmatorul caz
# definim o clasa si cream 1 obiect
class Car:
    color = 'orange'
    power = '100 hp'
````

````python
# definim un nou obiect masina
car_1 = Car()
car_1.color = 'green'

print(car_1.color)
print(Car.color)
# dupa cum putem vedea culoarea s-a schimbat
````

````python
# la primul obiect mai putem sa adaugam atribute
car_1.doors = 5
print('masina 1, nr. usi: ', car_1.doors)

# dar dupa cum putem vedea noul atribut nu este adaugat si in definitia initiala a obiectului
print(Car.doors)
````

````python
# efectul de shadowing, daca stergem un atribut al clasei
del car_1.color

print(car_1.color)
# cu toate ca noi am sters culoarea atribuita, revine la culoarea de baza a clasei
````

````python
# hai sa stergem un atribut adaugat suplimentar

del car_1.doors

print(car_1.doors)
# dupa cum putem vedea nu exista un atribut standar la care sa faca default
````

````python
# redefinim masina 1 cu o culoare, putere si nr. de usi si vedem ce atribute are obiectul
car_1 = Car()
car_1.color = 'green'
car_1.power = '200 hp'
car_1.doors = 5
car_1.__dict__
````

````python
# daca vrem sa vedem ce atribute are initial obiectul masina
car_1.__class__.__dict__
````

````python
# daca stergem un atribut din noul obiect, acesta va prelua valuarea din obiectul initial,
# in cazul in care exista
del car_1.color
print(car_1.__dict__)
print(car_1.__class__.__dict__)
print(car_1.color)
````

````python
# nu putem sa stergem color initial
del car_1.color
````

# Inheritance and composition

````python
class A:
    atrib_1 = 123

class B(A):
    pass

# clasa B a mostenit clasa A
B.atrib_1
````

````python
# putem sa folosim si metodele definite in clasa pe care o mostenim
# clasa A este clasa de baza, pentru ca ea este mostenita de B
class A:
    def count(self, n):
        for i in n:
            print(i)

            
class B(A):
    def __init__(self):
        n = range(5)
        self.count(n)
        
B()
````

# Accesarea unei clase de baza

````python
# in momenutl in care nu specificam o clasa de baza, python include de la el o clasa "object"
# daca nu definima o clasa de baza nu este necesar sa punem parantezele

# doua moduri de a scrie aceasi clasa
class A(object):
    pass

class A:
    pass
````

````python
# un exemplu de accesare datelor din clasa de baza

class Blog:
    def __init__(self, publisher, age):
        self.publisher = publisher
        self.age = age

class Article(Blog):
    def __init__(self, title, publisher, pages):
        self.title = title

# vom primi o eroare pentru ca cu toate ca clasa Article mosteneste clasa Blog, Blog nu a primit informatiile necesare
first_article = Article('Titlu', 5, 'Alex', 30)
first_article.publisher
````

````python
class Article(Blog):
    def __init__(self, title, pages, publisher, age):
        Blog.__init__(self, publisher, age)
        self.title = title

# ii spunem lui Python sa ruleze metoda __init__ a clasei Blog si ii punem argumentul self,
# pentru a o lega de obiectul curent

first_article = Article('Titlu', 5, 'Alex', 30)
first_article.publisher
````

````python
# daca modificam logica din Blog, atunci si clasa Article se va updata

# cum putem scrie mai frumos ce este scris mai sus
class Article(Blog):
    def __init__(self, title, pages, publisher, age):
        super(Article, self).__init__(publisher, age)
        self.title = title

# super este o functie care delega call-uri catre o clasa de baza
# avantajul functiei super este ca acuma putem sa modificam si numele clasei Blog pentru ca functia "super" se ocupa
# de call-urile catre clasa de baza
first_article = Article('Titlu', 5, 'Alex', 30)
first_article.publisher
````

# Method resolution order (MRO) is the order in which base classes are searched for a member during lookup.

````python
# acuma vom arata cum python cauta in clase atributul dorit si cum putem verifica
class A:
    atrib = 'a'

class B(A):
    atrib = 'b'

class C(A):
    atrib = 'c'

class D(B,C):
    pass

d = D()

# si B si C mostenesc pe A si D mosteneste pe B si C
d.atrib
````

````python
# prin functia MRO vedem ordinea in care obiectul cauta atributul si afiseaza ce gaseste prima data
# daca eliminam atrib din B, va afisa c
d.__class__.mro()
````

# Metode statice in clasa
@staticmethod means: when this method is called, we don't pass an instance of the class to it (as we normally do with methods). This means you can put a function inside a class but you can't access the instance of that class (this is useful when your method does not use the instance).

````python
# namespace-ul Wifi ne permite sa grupam mai multe metode de acelasi tip
class Wifi:
    @staticmethod
    def check_if_internet(wifi_name, wifi_pass):
        # stuff
        return True
    
    @staticmethod
    def reset_to_default():
        # get info from DB and reset internet
        return True
````

# Class methods

@classmethod means: when this method is called, we pass the class as the first argument instead of the instance of that class (as we normally do with methods). This means you can use the class and its properties inside that method rather than a particular instance.

````python
# Class methods sunt diferite fata de cele statice prin faptul ca primesc un argument initial cls
# classmethod foloseste cls care contine toata clasa, nu doar obiectul instantiat
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    @classmethod
    def from_tuple(cls, coords):
        # asterixul este de la *args adica ca pot trimite mai multe argumente sub forma de tulip, respectiv (x,y) (3,7)
        return cls(*coords)
    
    @classmethod
    def from_point(cls, point): # cls is Point
        return cls(point.x, point.y)

    
# este de recomandat sa folosim cls si self pentru ca este conventie
p = Point.from_tuple((3, 7))
print(p.x, p.y) # 3 7

q = Point.from_point(p)
print(q.x, q.y) # 3 7
````

# Property decorator

@property ne ajuta sa modificam proprietatile unui obiect, precum setter sau delete

````python
class Person:
    def __init__(self, n):
        self.age = n # oricine poate modifica aceasta valuare

obj = Person(10)
print(obj.age)

obj = Person(20)
print(obj.age)
````

````python
class SemiPrivatePerson:
    def __init__(self, n):
        self._age = n
    
    def get_age(self):
        return self._age
    
    def set_age(self, n):
        if 18 <= n <= 99:
            self._age = n
        else:
            raise ValueError('Age must be within [18, 99]')
````

````python
obj = SemiPrivatePerson(10)
print(obj.get_age())
````

````python
obj.set_age(20)
print(obj.get_age())
````

````python
obj._age = 21
print(obj.get_age())
# dupa cum putem vedea, putem modifica foarte usor valuarea obiectului folosind 
# getter and setter-ul obiectului builtin
````

````python
# folosim decoratorul Property pentru a putea face o copie a metodelor de baza ale 
# obiectului si sa putem interveni asupra acestora
class PythonPerson:
    def __init__(self, n):
        self._age = n
    # property este builtin si ne da acces la proprietatile unei clase
    @property
    def age(self):
        return self._age
    # @age este de fapt functia age de mai sus dar careia ii modificam setter-ul, ca sa nu poata oricine
    # sa modifice valuarea
    
    # cand vrem sa facem setter, adica sa atribuim o valuare lui _age, mai intai verificam inputul primit "n"
    # daca "n" nu se incadreaza returnam Value Error
    # daca "n" se incadreaza atunci putem interveni asupra lui _age, care este definit 
    @age.setter
    def age(self, n): # numele functiei trebuie sa fie exact ca cel mai de sus de la def age(self)
        if 18 <= n <= 99:
            self._age = n
        else:
            raise ValueError('Numarul trebuie sa fie intra [18, 100]')
    
    # putem implementa si restul de metode precum delete-ul, 
    @age.deleter
    def age(self):
        print('nu vreau sa il sterg')
````

````python
person = PythonPerson(39)
print(person.age)
````

````python
person.age = 42
print(person.age) # 42
````

````python
del(person.age)
````

````python
person.age = 100
````
