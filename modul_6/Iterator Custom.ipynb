```python
class OddEven:
    
    def __init__(self, data):
        self._data = data
        self.indexes = (list(range(0, len(data), 2)) + list(range(1, len(data), 2)))
        
        print(list(range(0, len(data), 2)))
        print(list(range(1, len(data), 2)))
        
        print(self.indexes)
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.indexes:
            return self._data[self.indexes.pop(0)]
        raise StopIteration
        

oddeven = OddEven('Aloha')
it = iter(oddeven) # aceasta cheama metoda: oddeven.__iter__

print(next(it))
print(next(it))
print(next(it))
print(next(it))
print(next(it))
```
