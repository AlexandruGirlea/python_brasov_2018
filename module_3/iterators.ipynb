## An object capable of returning its members one at a time.

# The itertools library

```python
from itertools import count

# count starts at 5 and adds 3 indefinitely.
# Use break to stop this otherwise infinite loop.
for n in count(5, 3):
    if n > 20:
        break
        
    print(n, end=', ')
```

```python
from itertools import compress

data = range(10)

print(list(data))
```

```python
even_selector = [1, 0] * 10
odd_selector = [0, 1] * 10

print(even_selector)

print(odd_selector)
```

```python
# compress selects items from data using the true and false values in the selector.
even_numbers = list(compress(data, even_selector))
odd_numbers = list(compress(data, odd_selector))

print(even_numbers)
print(odd_numbers)

# Although data has 10 items and the selector has 20,
# compress stops when the shorter input is exhausted.
```

```python
a = compress('ABCDEF', [True,True,1,0,1,1])
list(a)
```

```python
# Permutations
from itertools import permutations

print(list(permutations('ABC')))
```

```python
from itertools import combinations

lottery_numbers = []
for a in range(1,50):
    lottery_numbers.append(a)

a = combinations(lottery_numbers, 6)

a = list(a)

len(a)
```
