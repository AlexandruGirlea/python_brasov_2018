# Built-in containers
tuple, list, set, dict


"Containers" sunt acele tipuri de obiecte care pot contine un numar arbitrar de alte obiecte

# Tuple 
(immutable)

```python
# tuplele se definesc cu paranteze rotunde
a = (1,2,3,4)
a
```

```python
# putem extrage informatii din tuple; elementul de la pozitia 0
a[0]
```

```python
# asa verificam tipul
type(a)
```

# List
(mutable)

```python
# listele se definesc cu paranteze patrate
a = [1,2,3,4]

# elementul de la pozitia 0 se extrage la fel ca mai sus
a[0]
```

```python
# verificam tipul
type(a)
```

```python
# putem sa inlocuim valuarea elementului de la pozitia 0
a[0] = 5
a
```

# Set
(mutable)

```python
# definim o lista simpla
a = [1,1,3,4,9,5,6]

# set creaza un obiect de tip set care ordoneaza elementele din lista si le elimina pe cele duplicate
b = set(a)
b
```

```python
# putem adauga un element nou '7'
b.add(7)

# dupa cum putem vedea elementele sunt tot ordonate
b
```

```python
# nu putem adauga elemente deja existente
b.add(2)
b
```

# Dictionaries (Mapping types)
(mutable)

```python
# dictionarele se definesc cu acolade
a = {}
a['A'] = 1
a['B'] = 3

# dictionarele contin elemente de tip cheie : valuare, cheia este 'A' valuarea este 1
a
```

```python
# .keys() este o functie care ne permite sa vedem ce chei contine dictionarul 'a'
a.keys()
```
