Tipuri:

# namedtuple() Factory Function for Tuples with Named Fields

```python
from collections import namedtuple

point_tuple = namedtuple('Point', ['x', 'y'])

p = point_tuple(10, 11)

# ne arata obiectul creat
print(p)
```

```python
# ne arata numele clasei
print(p.__class__)
```

```python
# putem vedea codul din spate, care a fost generat pentru aceasta clasa
print(p._source)
```

```python
# extragem informatia
p.x + p.y
```

# deque (double-ended queue)

```python
# exemplu de lista simpla
a = [1,2,3,4]

#adaugam un element in partea dreapta '10'
a.append(10)
a
```

```python
# nu putem adauga in partea stanga folosind listele simple
a.appendleft(11)
a
```

```python
# folosim colectiile 'deque'
from collections import deque

d = deque(a)
d
```

```python
# acum putem adauga elemente noi si la stanga si la dreapta listei
d.append(12)
d.appendleft(100)
d
```

# ChainMap

```python
from collections import ChainMap

default_connection = {'host': 'localhost', 'port': 8000}
connection = {'port': 8080}

conn = ChainMap(connection, default_connection)
conn
```

```python
conn['port']
```

```python
# dupa cum putem vedea trece la valuarea default, pentru ca host nu este definit
conn['host']
```

```python
# daca il definim, el este adaugat la primul dictionar
conn['host'] = 'www.google.com'
conn
```

```python
conn['host']
```

# Counter

```python
from collections import Counter

# numaram de cate de comun este un element si le printam pe primele 3 cele mai comune
Counter('abracadabra').most_common(3)
```

```python
# exemplu nr.2 cu lista de numere
Counter([1,23,123,12,1,3,4,5,12,1,23]).most_common(3)
```

```python
# exemplu de cum putem interactiona cu obiectele de tip Counter
c = Counter(a=3, b=1)
d = Counter(a=1, b=2)
c+d
```

# OrderedDict

```python
# 'OrderedDict' il folosim pentru ordonarea dictionarelor
from collections import OrderedDict


d = {'banana': 3, 'apple': 4, 'pear': 1, 'orange': 2}
OrderedDict(sorted(d.items()))
```

# Defaultdict

```python
# 'defaultdict' putem sa il folosim pentru a crea un obiect gol care poate prelua informatii pe parcursul programului
# chiar si pentru chei inexistente
from collections import defaultdict

d = defaultdict(int)
d['A'] += 1
d
```

```python
# un dictionar simplu nu poate face acest lucru cu operatorii de tip +=
a = {}
a['A'] += 1
a
```
