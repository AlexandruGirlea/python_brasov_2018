# Importurile in Python sunt de 2 tipuri:
- Absolute
- Relative

Un import absolut foloseste calea completa catre modulul pe care dorim sa il importam

Un import relativ se foloseste de calea modulului curent pentru a importa noi module.

```python
# libraria "imp" se ocupa de importurile in python
import imp

# 'os' este un modul in python cu ajutorul caruia putem rula anumite comenzi de sistem
# acum vom vedea de este localizat modulul 'os'
imp.find_module('os')
```

```python
# in cele ce urmeaza vom folosi modulul 'os'
import os

# functia "getcwd" este pentru a obtine directorul curent in care lucram
os.getcwd()
```
