## Obiectele a caror valoare se poate schimba se numesc "mutable"

## Obiectele a caror valoare nu se schimba se numesc "immutable"

```python
# a este un tip de obiect a carui valoare nu se schimba
a = 30

# locatia in memorie a lui 'a', este
id(a)
```

```python
# daca schimbam valoarea lui 'a'
a = 31

# acuma 'a'se afla la alta locatie in memorie ceea ce inseamna ca obiectul de la locatia anterioara '10920352'
# nu s-a schimbat
id(a)
```

Valorile de tip integer sunt "immutable", am schimbat numai label-ul (am asociat "a" cu 31)

```python
# construim un obiect Persoana care are un singur atribut, varsta (age)
class Person:
    age = 0
    def __init__(self, age):
        self.age = age
```

Acesta este un exemplu de obiect care poate sa isi schimbe valoarea

```python
# ii atribuim obiectului Andrei varsta de 30 de ani
Andrei = Person(age=30)
Andrei.age
```

```python
# 'Andrei' se afla la aceasta locatie in memorie
id(Andrei)
```

```python
# daca ii schimbam varsta lui 'Andrei'
Andrei.age = 27
Andrei.age
```

```python
# obiectul 'Andrei' se va afla la aceeasi locatie in memorie dar va avea alta varsta
id(Andrei)
```

Dupa cum se poate observa obiectul este de tip "mutable", a preluat alta valoare dar este se afla la acelasi "id" in memorie
