# While loop

For loop-ul este folosind pentru obiecte finite, iar While loop este folosit pentru ca un program sa ruleze pana cand o conditie este indeplinita.

```python
n = 5
while n > 0:
    print(n)
    n -= 1
```

# Break and continue statements

```python
from datetime import date, timedelta
today = date.today()
tomorrow = today + timedelta(days=1) # today + 1 day is tomorrow
products = [
    {'sku': '1', 'expiration_date': today, 'price': 100.0},
    {'sku': '2', 'expiration_date': tomorrow, 'price': 50},
    {'sku': '3', 'expiration_date': today, 'price': 20},
]
for product in products:
    if product['expiration_date'] != today:
        # continue opreste iteratia curenta si trece la urmatoarea
        continue
    product['price'] *= 0.8 # equivalent to applying 20% discount
    print('Price for sku', product['sku'],'is now', product['price'])
```

```python
items = [0, False, None, 0.0, True, 0, 1, 2]

found = False

for item in items:
    print('scanning item', item)
    if item:
        found = True # we update the flag
        # break opreste iteratia complet
        break

found
```

```python
if found: # we inspect the flag
    print('At least one item evaluates to True')
else:
    print('All items evaluate to False')
```

# Special "else" statement in "for" loops

```python
# definim o clasa pentru exceptie
class MyError(Exception):
    def __init__(self, value):
        self.value = value
    def __str__(self):
        return repr(self.value)
```

```python
people = [('Alex', 17), ('Ovidiu', 9), ('Andrei', 13), ('Octavian', 8)]

driver = None

for person, age in people:
    if age >= 18:
        driver = (person, age)
        break

if driver is None:
    raise MyError('Driver not found.')
```
