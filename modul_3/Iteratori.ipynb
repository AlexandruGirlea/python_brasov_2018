## An object capable of returning its members one at a time.

# Libraria Itertools

```python
from itertools import count

# count incepe de la 5 si adauga + 3 la infinit
# count ne creaza un infinite loop si trebuie sa folosim break
for n in count(5, 3):
    if n > 20:
        break
        
    print(n, end=', ')
```

```python
from itertools import compress

data = range(10)

print(list(data))
```

```python
even_selector = [1, 0] * 10
odd_selector = [0, 1] * 10

print(even_selector)

print(odd_selector)
```

```python
# compress itereaza asupra obiectului "data" in functie de lista "even_selector" care este compusa din elemente True si False
even_numbers = list(compress(data, even_selector))
odd_numbers = list(compress(data, odd_selector))

print(even_numbers)
print(odd_numbers)

# OBS: chiar daca "data" este lung de 10 elemente, iar selectorul par este de 20 elemente, iteratorul "compress"
# se va opri imediat ce obiectul "data" s-a terminat.
```

```python
a = compress('ABCDEF', [True,True,1,0,1,1])
list(a)
```

```python
# combinatii
from itertools import permutations

print(list(permutations('ABC')))
```

```python
from itertools import combinations

loto_numbers = []
for a in range(1,50):
    loto_numbers.append(a)

a = combinations(loto_numbers, 6)

a = list(a)

len(a)
```
