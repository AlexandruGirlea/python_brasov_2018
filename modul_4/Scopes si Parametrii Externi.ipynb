## Why use functions?
Functions are among the most important concepts and constructs of any language,
so let me give you a few reasons why we need them:
- They reduce code duplication in a program. By having a specific task
taken care of by a nice block of packaged code that we can import and
call whenever we want, we don't need to duplicate its implementation.
- They help in splitting a complex task or procedure into smaller blocks,
each of which becomes a function.
- They hide the implementation details from their users.
- They improve traceability.
- They improve readability.

https://docs.python.org/2/library/functions.html

```python
def order(*args, **kwargs):
    if args:
        for i in args:
            print(i)
    if kwargs:
        for i in kwargs:
            print(i)
        print(kwargs)
order('Alex', 'Adi', 'Andrei')

# dictionarele nu au ordine
order(key_1='abcd', key_2='abcd') 
```

# Scopes
Scope types:
- local;
- enclosing;
- global;
- built-in.

```python
test = 0

def function_1():
    test = 1
    print('test este egal cu: ', test)

function_1()
```

```python
test
```

```python
def outer():
    test = 1 # outer scope
    def inner():
        test = 2 # inner scope
        print('inner:', test)
    inner()
    print('outer:', test)
    
test = 0 # global scope
outer()

print('global:', test)
```

# Folosirea Parametrilori Externi
Principii:
- folosirea unui parametru extern asigneaza un obiect unei variabile locale

```python
x = 2

def func(y):
    print(y)
func(x)
```

- asignarea unei valori unui variabile, se face numai local si nu afecteaza variabila globala

```python
x = 2
def func(x):
    x = 7

func(x)
print(x)
```

- functiile pot modifica obiectele "mutable"

```python
a = [1,2,3,4]

def func():
    a[0] = 10

func()
a
```

```python
a = [1,2,3,4]

def func():
    a[0] = 10
    a = 'acesta este un string'

func()
print(a)
# primim o eroare pentru ca variabila locala (a) intra in conflict cu variabila globala (a)
```

```python
a = [1,2,3,4]

def func(a):
    a[0] = 10
    a = 'acesta este un string'
    print('varibial a din interiorul functiei: ', a)

func(a)
print(a)
# dupa cum se poate vedea variabiala globala a fost modificata iar variabila locala a are o alta valuare
```

## Tipuri de parametrii externi:
- Positional arguments (cei mai comuni)
- Keyword arguments si valorile default
- Variable positional arguments
- Variable keyword arguments
- Keyword-only arguments

1. Positional arguments (cei mai comuni)

```python
def func(a, b, c):
    print(a, b, c)

func(1,2,3)
```

2. Keyword arguments si valorile default
(Keyword arguments are assigned by keyword using the name=value syntax.)

```python
func(a=1, c=2, b=3)
```

```python
# folosirea valorilor default
def func(a, b=10, c=11):
    print(a, b, c)
    
func(1)
func(a=1, b=2, c=3)
```

3. Variable positional arguments -> * face unpacking a valorilor

```python
# functie pentru determinarea minimei unei tuple
def minimum(*n):
    # print(n) # n is a tuple
    if n: # explained after the code
        minim = n[0]
        for value in n[1:]:
            if value < minim:
                minim = value
        print(minim)
            
minimum(1,3,4,5,-1)
```

```python
n = (1,3,4,5,-1)
n
```

```python
n[1:]
```

```python
n[:-1]
```

4. Variable keyword arguments -> syntax ( ** instead of * )

```python
def func(**ioi):
    print(ioi)

func()
```

```python
func(a=1, b=2)
```

```python
a = {'a':123}
a.get('b')
```

5. Keyword-only arguments (se pot defini fie dupa un "variable keyword" sau dupa un simplu *)

```python
def func(*a, b):
    print(a, b)

func(1,2,3, b=4)
```

```python
func(b=4)
```

```python
func(1,2,3,4)
```

```python
def func(*, b):
    print(b)

func(1)
```

```python
func(b=1)
```
