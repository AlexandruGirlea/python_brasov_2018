### De obicei functiile returneaza o singura valuare dar python poate returna o tuple

```python
# obs daca o functie nu returneaza nimic, aceasta va returna by default None

def func():
    pass

a = func()

print(a)
```

```python
def func(n):
    if n in [1,2,3]:
        return n
    else:
        return 0

func(1)
```

```python
func(5)
```

```python
from functools import reduce
from operator import mul

# tema de casa, cum functioneaza aceasta functie (de trimis pe mail)
def factorial(n):
    return reduce(mul, range(1, n + 1), 1)

f5 = factorial(5)

f5
```

```python
# o functie in python poate returna si rezultatul unei alte functii
def func(*n):
    return min(n)

func(1,3,4,-7)
    
```

```python
# returnarea multiplelor valori
def two_value_function():
    return 1,False

a, b = two_value_function()
a
```

# Principii de baza ale functiilor:
- Functiile trebuie sa faca un singur lucru
- Functiile trebuie sa fie scurte
- Functiile trebuie sa primeasca cat mai putini parametrii
- Trebuie luat in considerare ce returneaza o functie
- Functiile nu trebuie sa aiba efecte secundare
