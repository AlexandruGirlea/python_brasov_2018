### Atunci cand o functie se cheama pe ea insasi sa produca un rezultat, aceasta este recursiva

```python
def factorial(n):
    if n in (0, 1): # base case
        return 1
    return factorial(n - 1) * n # recursive case

factorial(4) # 1 x 2 x 3 x 4
```
