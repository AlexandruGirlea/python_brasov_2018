## Every function is a fully-fledged object and, as such, they have many attributes.

```python
def multiplication(a, b=2):
    return a * b
```

```python
# b este by default 2
# analizam atributele unei functii
multiplication.__defaults__
```

```python
multiplication.__code__
```

```python
special_attributes = dir(multiplication)
special_attributes
```

```python
# extragem atributele
for attribute in special_attributes:
    print('\n')
    print(attribute, '->', getattr(multiplication, attribute))
```

```python
# lista functiilor built in
dir(__builtins__)
```

```python
# built in ZIP function

a = (0,1,2,3,4)
b = (10,11,12,13, 14, 15)

list(zip(a,b))
```

```python
# built in TYPE
type(a)
```

```python
# built in RANGE
c = range(0,10)
c
```

```python
list(c)
```
