## These functions are called lambdas in Python

Functiile anonime nu sunt legate de un identificator

```python
def func(a, b):
    return(a + b)

a = func(10, 11)
a
```

```python
a = lambda x, y: x + y

a(10, 11)
```

```python
# functia "map" primeste o functie ca parametru initial si o lista
# "map" aplica functia pentru fiecare element din lista

def func(x):
    return x * 2

map(func, [1,2,3])
```

```python
list(map(func, [1,2,3]))
```

```python
# o singura linie de cod, pentru o functie pe care o folosesc doar aici
list(map(lambda x : x*2, [1, 2, 3]))
```

```python
list_1 = [1,2,3,10]

newList = filter(lambda x: x / 2 == 5, list_1)
for i in newList:
    print(i)
```

```python
a = lambda x: x + 1
a(1)
```

```python
def func(n):
    return n+1
func(1)
```

```python
a = lambda x: x / 2 == 5
a(10)
```
