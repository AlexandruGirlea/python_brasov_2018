# Unit testing
Este acel mod de a scrie teste care verifica regiunile de cod pe rand.

Un test, este o sectiune de cod al carui scop este sa verifice o portiune a sistemului nostru.

Partile componenete ale unui test sunt:

- pregatirea datelor pentru testare
- executi putin cod pentru cu datele pregatite
- verifici rezultatul (ex: o funtie returneaza ce trebuie, faptul ca s-au scris niste date in DB, etc.)

Idei principale pentru scrierea testelor:
- testele trebuie sa fie cat mai simple (scrise cat mai lizibil)
- testul trebuie sa verifice un singur lucru
- testele nu trebuie sa faca presupuneri ca anumite informatii sunt corecte, si trebuie sa testeze exact ce primesc fara sa altereze informatiile
- testele trebuie sa verifice ce trebuie sa faca o functie si nu cum o face
- testele trebuie sa fie rapide si sa foloseasca foarte putine resurse

## Acum vom crea un test mai complex

````python
# cream un obiect Student
class Student:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
    
    # ii atribuim proprietatea get_full_name
    # ceea ce inseamna ca nu mai trebuie sa rulam functia ca get_full_name()
    # acuma putem rula functia direct a un atribut .get_full_name
    @property
    def get_full_name(self):
        return '{} {}'.format(self.first_name, self.last_name)
    
    @property
    def get_all_data_on_student(self):
        return 'First name: {}, Last name: {}, age is: {}'.format(self.first_name, self.last_name, self.age)
````

````python
# incepem testul
import unittest


class TestStudent(unittest.TestCase):
    # aceasta functie ruleaza prima data cand rulam un test, de aceea ne cream obiectele aici
    def setUp(self):
        print('setUp')
        self.student_1 = Student(first_name='Alex', last_name='Girlea', age=30)
        self.student_2 = Student(first_name='Damian', last_name='Folvaiter', age=31)
        
    # aceasta functie ruleaza la sfarsitul fiecarui test
    def tearDown(self):
        print('tearDown')
    
    def test_get_full_name(self):
        print('test_get_full_name')
        self.assertEqual(self.student_1.get_full_name, 'Alex Girlea')
        self.assertEqual(self.student_2.get_full_name, 'Damian Folvaiter')
    
    def test_get_all_data_on_student(self):
        print('test_get_all_data_on_student')
        self.assertEqual(self.student_1.get_all_data_on_student, 'First name: Alex, Last name: Girlea, age is: 30')
        self.assertEqual(self.student_2.get_all_data_on_student, 'First name: Damian, Last name: Folvaiter, age is: 31')
````

````python
# dupa cum putem vedea setUp a rulat prima data si dupa aceea testul si dupa aceea tearDown care poate sa fie
# o metoda de curatare

# ce mai trebuie sa observam este ca testele nu au fost executate de sus in jos
unittest.main(argv=['first-arg-is-ignored'], exit=False)
````

## Exemplu de test folosind chemarea unei exceptii

````python
# resetam jupyter notebook
%reset
````

````python
# cream un obiect Student
class Student:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        if type(age) != int:
            raise TypeError
        self.age = age
    
    # ii atribuim proprietatea get_full_name
    # ceea ce inseamna ca nu mai trebuie sa rulam functia ca get_full_name()
    # acuma putem rula functia direct a un atribut .get_full_name
    @property
    def get_full_name(self):
        return '{} {}'.format(self.first_name, self.last_name)
    
    @property
    def get_all_data_on_student(self):
        return 'First name: {}, Last name: {}, age is: {}'.format(self.first_name, self.last_name, self.age)
    
    def check_age_limit(self,n):
        if type(n) != int:
            raise TypeError
        else:
            # do some stuff
            pass
````

````python
import unittest

class TestStudent(unittest.TestCase):
    # aceasta functie va rula prima data si doar o singura data (ex: importare date din baza de date)
    @classmethod
    def setUpClass(cls):
        print('\n classSetUp')
    
    # aceasta functie va rula o singura data la sfarsitul testului
    @classmethod
    def tearDownClass(cls):
        print('\n tearDownClass')
    
    # aceasta functie va rula prima data la fiecare test
    def setUp(self):
        print('\n setUp', '-'*10)
        self.student_1 = Student(first_name='Alex', last_name='Girlea', age=30)
        self.student_2 = Student(first_name='Damian', last_name='Folvaiter', age=31)
        
    # aceasta functie ruleaza la sfarsitul fiecarui test
    def tearDown(self):
        print('\n tearDown', '_'*10)
    
    def test_get_full_name(self):
        print('\n test_get_full_name')
        self.assertEqual(self.student_1.get_full_name, 'Alex Girlea')
        self.assertEqual(self.student_2.get_full_name, 'Damian Folvaiter')
    
    def test_get_all_data_on_student(self):
        print('\n test_get_all_data_on_student')
        self.assertEqual(self.student_1.get_all_data_on_student, 'First name: Alex, Last name: Girlea, age is: 30')
        self.assertEqual(self.student_2.get_all_data_on_student, 'First name: Damian, Last name: Folvaiter, age is: 31')
    
    def test_check_age_limit(self):
        print('\n check_age_limit')
        # verificam daca un string returneaza un type error
        # daca returneaza TypeError atunci testul trece
        self.assertRaises(TypeError, self.student_1.check_age_limit, '123')
                        
````

````python
unittest.main(argv=['first-arg-is-ignored'], exit=False)
````
