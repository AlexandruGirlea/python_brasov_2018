# The collections module

This module provides specialized containers for situations where `tuple`, `list`, `set`, and `dict` are not enough.

Types: `namedtuple`, `deque`, `ChainMap`, `Counter`, `OrderedDict`, `defaultdict`, `UserDict`, `UserList`, `UserString`.

More information: https://docs.python.org/3/library/collections.html

# namedtuple() Factory Function for Tuples with Named Fields

```python
from collections import namedtuple

point_tuple = namedtuple('Point', ['x', 'y'])

p = point_tuple(10, 11)

# Display the newly created object.
print(p)
```

```python
# Display its class.
print(p.__class__)
```

```python
# Display the generated class source (available in the original course environment).
print(p._source)
```

```python
# Access the stored values.
p.x + p.y
```

# deque (double-ended queue)

```python
# A simple list
a = [1,2,3,4]

# Append the value 10 on the right.
a.append(10)
a
```

```python
# A list has no appendleft() method.
a.appendleft(11)
a
```

```python
# Use a deque from collections.
from collections import deque

d = deque(a)
d
```

```python
# Now we can append items at either end.
d.append(12)
d.appendleft(100)
d
```

# ChainMap

```python
from collections import ChainMap

default_connection = {'host': 'localhost', 'port': 8000}
connection = {'port': 8080}

conn = ChainMap(connection, default_connection)
conn
```

```python
conn['port']
```

```python
# Fall back to the default value because host is absent from the first mapping.
conn['host']
```

```python
# Assigning a value adds it to the first mapping.
conn['host'] = 'www.google.com'
conn
```

```python
conn['host']
```

# Counter

```python
from collections import Counter

# Count occurrences and show the three most common items.
Counter('abracadabra').most_common(3)
```

```python
# Another example, using a list of numbers
Counter([1,23,123,12,1,3,4,5,12,1,23]).most_common(3)
```

```python
# Combine Counter objects.
c = Counter(a=3, b=1)
d = Counter(a=1, b=2)
c+d
```

# OrderedDict

```python
# Use OrderedDict to retain the order of the sorted dictionary items.
from collections import OrderedDict


d = {'banana': 3, 'apple': 4, 'pear': 1, 'orange': 2}
OrderedDict(sorted(d.items()))
```

# Defaultdict

```python
# A defaultdict supplies a default value for missing keys,
# so we can accumulate values without initializing each key.
from collections import defaultdict

d = defaultdict(int)
d['A'] += 1
d
```

```python
# An ordinary dictionary raises KeyError when += accesses a missing key.
a = {}
a['A'] += 1
a
```
