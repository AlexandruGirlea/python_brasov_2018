## Objects whose values can change are mutable.

## Objects whose values cannot change are immutable.

```python
# a refers to an immutable object.
a = 30

# Inspect its identity.
id(a)
```

```python
# Assign a different value to a.
a = 31

# Now a refers to a different object;
# the original object has not changed.
id(a)
```

Integers are immutable. We have rebound the name `a` to a different value, `31`.

```python
# Create a Person class with one attribute: age.
class Person:
    age = 0
    def __init__(self, age):
        self.age = age
```

This is an example of an object whose state can change.

```python
# Create a person named Andrei, aged 30.
Andrei = Person(age=30)
Andrei.age
```

```python
# Inspect the identity of Andrei.
id(Andrei)
```

```python
# Change the age of Andrei.
Andrei.age = 27
Andrei.age
```

```python
# Andrei has the same identity but a different age.
id(Andrei)
```

The object is mutable: its age has changed, while its identity remains the same.
