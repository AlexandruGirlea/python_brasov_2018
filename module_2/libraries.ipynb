# Two types of imports in Python

- Absolute imports use the full module path.
- Relative imports use the current package as a starting point.

```python
# The historical imp module handles imports in this example.
import imp

# The os module provides operating-system interfaces.
# Find where the os module is located.
imp.find_module('os')
```

```python
# Use the os module.
import os

# getcwd() returns the current working directory.
os.getcwd()
```
