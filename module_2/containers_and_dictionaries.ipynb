# Built-in containers

`tuple`, `list`, `set`, `dict`

Containers are objects that can hold other objects.

# Tuple 
(immutable)

```python
# Define a tuple using parentheses.
a = (1,2,3,4)
a
```

```python
# Retrieve the item at index 0.
a[0]
```

```python
# Inspect its type.
type(a)
```

# List
(mutable)

```python
# Define a list using square brackets.
a = [1,2,3,4]

# Retrieve the item at index 0 in the same way.
a[0]
```

```python
# Inspect its type.
type(a)
```

```python
# Replace the item at index 0.
a[0] = 5
a
```

# Set
(mutable)

```python
# Define a simple list.
a = [1,1,3,4,9,5,6]

# A set removes duplicate values; it does not guarantee sorted order.
b = set(a)
b
```

```python
# Add the value 7.
b.add(7)

# The set contains unique values, with no guaranteed order.
b
```

```python
# Add the value 2. Adding the same value again would not create a duplicate.
b.add(2)
b
```

# Dictionaries (Mapping types)
(mutable)

```python
# Define dictionaries using braces.
a = {}
a['A'] = 1
a['B'] = 3

# Dictionaries store key-value pairs: the key 'A' has the value 1.
a
```

```python
# keys() returns a view of the keys in dictionary a.
a.keys()
```
