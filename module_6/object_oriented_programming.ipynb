# Attribute shadowing

````python
# Attribute shadowing
# Define a class, then create an instance.
class Car:
    color = 'orange'
    power = '100 hp'
````

````python
# Create a new car instance.
car_1 = Car()
car_1.color = 'green'

print(car_1.color)
print(Car.color)
# The instance color differs from the class color.
````

````python
# Add an attribute to the instance.
car_1.doors = 5
print('car 1, number of doors: ', car_1.doors)

# The new attribute is not added to the class itself.
print(Car.doors)
````

````python
# Remove the instance attribute to demonstrate shadowing.
del car_1.color

print(car_1.color)
# Attribute lookup now falls back to the class color.
````

````python
# Delete the additional instance attribute.

del car_1.doors

print(car_1.doors)
# There is no class attribute to fall back to, so lookup raises an error.
````

````python
# Create a car with a color, power, and door count, then inspect its attributes.
car_1 = Car()
car_1.color = 'green'
car_1.power = '200 hp'
car_1.doors = 5
car_1.__dict__
````

````python
# Inspect the attributes on the original class.
car_1.__class__.__dict__
````

````python
# After deleting an instance attribute, lookup falls back to the class attribute
# if one exists.
del car_1.color
print(car_1.__dict__)
print(car_1.__class__.__dict__)
print(car_1.color)
````

````python
# We cannot delete the class attribute through the instance this way.
del car_1.color
````

# Using self

````python
# self is the conventional name for the current instance.

class Cube:
    attribute_1 = 3
    def calculate(self):
        return self.attribute_1 ** 3
````

# Inheritance and composition

````python
class A:
    attribute_1 = 123

class B(A):
    pass

# B inherits from A.
B.attribute_1
````

````python
# We can use methods defined in the class we inherit from.
# A is the base class of B.
class A:
    def count(self, n):
        for i in n:
            print(i)

            
class B(A):
    def __init__(self):
        n = range(5)
        self.count(n)
        
B()
````

### Composition

````python
class Legs:
    def walk(self):
        pass
    def run(self):
        pass

class Skin:
    pass

class Animal:
    def __init__(self):
        self.skin = Skin()
        self.legs = Legs()
        
# The difference between inheritance and composition:

# Inheritance: a subclass inherits attributes from its base class (a racing car is a kind of car).
# Composition: an object is built from other objects (a car has several parts).
````

# Accessing a base class

````python
# If no base class is specified, Python uses object.
# Parentheses are optional when there is no explicit base class.

# Two ways to write the same class:
class A(object):
    pass

class A:
    pass
````

````python
# Access data from a base class.

class Blog:
    def __init__(self, publisher, age):
        self.publisher = publisher
        self.age = age

class Article(Blog):
    def __init__(self, title, publisher, pages):
        self.title = title

# This call raises an error: the arguments do not match the Article initializer.
first_article = Article('Title', 5, 'Alex', 30)
first_article.publisher
````

````python
class Article(Blog):
    def __init__(self, title, pages, publisher, age):
        Blog.__init__(self, publisher, age)
        self.title = title

# Call the Blog initializer and pass self
# to initialize the current instance.

first_article = Article('Title', 5, 'Alex', 30)
first_article.publisher
````

````python
# Changes to initialization in Blog will also apply through this call.

# A cleaner way to write the previous example
class Article(Blog):
    def __init__(self, title, pages, publisher, age):
        super(Article, self).__init__(publisher, age)
        self.title = title

# super delegates method calls according to the method resolution order.
# It avoids hard-coding the base class name, so changing the base class
# does not require updating this initializer call.
first_article = Article('Title', 5, 'Alex', 30)
first_article.publisher
````

# Method resolution order (MRO) is the order in which base classes are searched for a member during lookup.

````python
# Inspect how Python searches base classes for an attribute.
class A:
    attribute = 'a'

class B(A):
    attribute = 'b'

class C(A):
    attribute = 'c'

class D(B,C):
    pass

d = D()

# B and C inherit from A; D inherits from B and C.
d.attribute
````

````python
# mro() shows the order in which classes are searched for attributes.
# Removing the attribute from B would make lookup return the value from C.
d.__class__.mro()
````

# Static methods
@staticmethod means: when this method is called, we don't pass an instance of the class to it (as we normally do with methods). This means you can put a function inside a class but you can't access the instance of that class (this is useful when your method does not use the instance).

````python
# Use the Wifi class to group related operations.
class Wifi:
    @staticmethod
    def check_if_internet(wifi_name, wifi_pass):
        # Example implementation
        return True
    
    @staticmethod
    def reset_to_default():
        # Retrieve configuration from a database and reset the connection.
        return True
````

# Class methods

@classmethod means: when this method is called, we pass the class as the first argument instead of the instance of that class (as we normally do with methods). This means you can use the class and its properties inside that method rather than a particular instance.

````python
# Class methods receive the class as their first argument, conventionally named cls.
# cls refers to the class rather than a particular instance.
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    @classmethod
    def from_tuple(cls, coords):
        # The asterisk unpacks the tuple into positional arguments, such as x=3 and y=7.
        return cls(*coords)
    
    @classmethod
    def from_point(cls, point): # cls is Point.
        return cls(point.x, point.y)

    
# Use cls and self to follow Python naming conventions.
p = Point.from_tuple((3, 7))
print(p.x, p.y) # 3 7

q = Point.from_point(p)
print(q.x, q.y) # 3 7
````

# Public and Private in Python

````python
# Python relies on conventions and name mangling rather than strictly private attributes.
# Take care when naming attributes shared by related classes.


# Define classes whose attribute names collide.
class A:
    def __init__(self, number):
        self.number = number
        
    def output_1(self):
        print('the number is: {}'.format(self.number))

class B(A):
    def output_2(self, number):
        self.number = number
        print('the number is: {}'.format(self.number))

obj = B(10)
````

````python
obj.output_1()
````

````python
obj.output_2(20)
````

````python
# The second method has also changed the value used by output_1.
obj.output_1()
````

````python
# Rewrite the classes using two leading underscores on the attribute names.
class A:
    def __init__(self, number):
        self.__number = number
        
    def output_1(self):
        print('the number is: {}'.format(self.__number))

class B(A):
    def output_2(self, number):
        self.__number = number
        print('the number is: {}'.format(self.__number))

obj = B(10)
````

````python
obj.output_1()
````

````python
obj.output_2(20)
````

````python
obj.output_1()
````

# Property decorator

`@property` lets us manage attribute access with getter, setter, and deleter methods.

````python
class Person:
    def __init__(self, n):
        self.age = n # This attribute can be changed directly.

obj = Person(10)
print(obj.age)

obj = Person(20)
print(obj.age)
````

````python
class SemiPrivatePerson:
    def __init__(self, n):
        self._age = n
    
    def get_age(self):
        return self._age
    
    def set_age(self, n):
        if 18 <= n <= 99:
            self._age = n
        else:
            raise ValueError('Age must be within [18, 99]')
````

````python
obj = SemiPrivatePerson(10)
print(obj.get_age())
````

````python
obj.set_age(20)
print(obj.get_age())
````

````python
obj._age = 21
print(obj.get_age())
# The attribute remains directly accessible;
# assigning to it bypasses the getter and setter methods.
````

````python
# Use a property to customize attribute access
# through getter, setter, and deleter methods.
class PythonPerson:
    def __init__(self, n):
        self._age = n
    # property is a built-in decorator for managed attributes.
    @property
    def age(self):
        return self._age
    # Add a setter to the age property defined above
    # to validate changes to its value.
    
    # Before assigning to _age, validate the input n.
    # Raise ValueError if n is outside the allowed range.
    # Otherwise, update _age.
    @age.setter
    def age(self, n): # The setter uses the same name as the property getter.
        if 18 <= n <= 99:
            self._age = n
        else:
            raise ValueError('Age must be within [18, 99]')
    
    # A property can also define a deleter.
    @age.deleter
    def age(self):
        print('deletion is disabled')
````

````python
person = PythonPerson(39)
print(person.age)
````

````python
person.age = 42
print(person.age) # 42
````

````python
del(person.age)
````

````python
person.age = 100
````
