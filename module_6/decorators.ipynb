## Reusing short, repetitive pieces of code

```python
# Example with ordinary functions

import time

def a():
    time.sleep(0.2)

def b():
    time.sleep(0.3)

initial_time = time.time()

a()

print('execution time for function a: ', time.time()-initial_time)


initial_time = time.time()

b()

print('execution time for function b: ', time.time()-initial_time)
```

```python
# Extract the timing code into a reusable function.

def time_measure(func):
    initial_time = time.time()
    func()
    print('Function', func.__name__, 'takes:', time.time() - initial_time)
```

```python
time_measure(a)
time_measure(b)
```

```python
# Now let functions a and b accept arguments.
def a(sleep_time):
    time.sleep(sleep_time)
    
def b(sleep_time):
    time.sleep(sleep_time)
```

```python
# This call raises an error: a(1) runs immediately and returns None,
# so time_measure receives None instead of a callable.
time_measure(a(1))
```

```python
# Define a nested function to forward arguments
# to the original function.
def time_measure(func):
    
    def wrapper(*args, **kwargs):
        initial_time = time.time()
        func(*args, **kwargs)
        print('Function', func.__name__, 'takes:', time.time() - initial_time)
    # Return the wrapper function, which accepts args and kwargs
    # and forwards them when it is called.
    return wrapper
```

```python
a = time_measure(a)
a(sleep_time=1)
```

```python
# time_measure is a decorator.
# Here is a cleaner way to write it.
def decorator(f):
    pass

@decorator
def a(arg_1):
    pass

# This is equivalent to the syntax above:
def a(arg_1):
    pass

a = decorator(a)
```

```python
# Apply multiple decorators.
@ decorator_1
@ decorator_2
def a(arg_1):
    pass

# This is equivalent to:
a = decorator_1(decorator_2(a))

# The decorator closest to the function is applied first.
# Here, decorator_1 receives the result of decorator_2.
```

```python
# A decorator with an argument, written as an explicit function call
def time_measure(func, arg_1):
    print(arg_1)
    
    def wrapper(*args, **kwargs):
        initial_time = time.time()
        func(*args, **kwargs)
        print('Function', func.__name__, 'takes:', time.time() - initial_time)
    # Return the wrapper function, which accepts args and kwargs
    # and forwards them when it is called.
    return wrapper

def a(sleep_time):
    time.sleep(sleep_time)

a = time_measure(a, arg_1='this is argument 1')
a(1)
```

```python
# Decorator syntax for passing an argument
@decorator(arg_1)
def a(sleep_time):
    time.sleep(sleep_time)
```

```python
# Use functools.wraps to preserve the original function metadata.
from functools import wraps

def time_measure(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        initial_time = time.time()
        func(*args, **kwargs)
        print('Function', func.__name__, 'takes:', time.time() - initial_time)
    # Return the wrapper function, which accepts args and kwargs
    # and forwards them when it is called.
    return wrapper
```

```python
@time_measure
def a(sleep_time):
    time.sleep(sleep_time)

a(1)
```

```python
# Function a has retained its original name.
a.__name__
```

```python
# A practical decorator example
def minimum_income(income):
    def decorator(func):
        @wraps(func)
        def wrapper(*args, **kwargs):
            result = func(*args, **kwargs)
            if result > income:
                print('income is sufficient')
            else:
                print('income is too low')
        return wrapper
    return decorator

@minimum_income(2000)
def income_calculator(x):
    return x * 2

income_calculator(100)
```
