```python
# Generators do not have to collect all results before returning them.
# They yield one value, pause, and resume to produce the next value.
```

```python
 # classic function approach
def get_squares(n):
    return [x ** 2 for x in range(n)]
print(get_squares(3))
```

```python
def get_squares_gen(n): # generator approach
    for x in range(n):
        yield x ** 2 # we yield, we don't return
        
a = get_squares_gen(3)

print(next(a))
print(next(a))
# a generator will instead raise a StopIteration exception
```

```python
# Generators are useful when the data is too large to store in a list.
```

```python
# next(generator) calls the __next__ method shown below.
a.__next__()
```

```python
# Generators also provide send, throw, and close methods.

# send passes a value back into the generator.
# throw raises an exception inside the generator.
# close closes the generator.
```

```python
stop = True
def count(n=0):
    while stop:
        yield n
        n += 1
a = count()
print(a.__next__())
print(a.__next__())
print(a.__next__())
stop = False
try:
    count().__next__()
except StopIteration:
    print('exception raised StopIteration')
```

```python
# Use send().
def count(start=0):
    n = start
    while True:
        result = yield n
        print(type(result), result)
        if result == 'quit':
            break
        n += 1
        
c = count()
print(next(c))
print(c.send('hello'))
print(next(c))
try:
    print(c.send('quit'))
except StopIteration:
    print('exception raised StopIteration')
```

## The yield from expression

```python
# we yield one at a time
def count(start, end):
    for n in range(start, end):
        yield n + 1

for n in count(2, 5):
    print(n)
```

```python
# yield from g is == for i in j: yield i
def count(start, end):
    yield from (n+1 for n in range(start, end))

for n in count(2, 5):
    print(n)
```

## Generator expressions

```python
# Use parentheses instead of square brackets to create a generator expression.
squares_list = [k**2 for k in range(5)]
squares_list
```

```python
squares_list = (k**2 for k in range(5))
squares_list
```

```python
print(next(squares_list))
print(next(squares_list))
print(next(squares_list))
```

```python
# if this was not a generator it will die
s = sum(n**2 for n in range(10**8))
```

```python
# The generator lets us compute the sum without storing all the squared values.
s
```

### Practical principles

- Prefer a generator when values can be processed one at a time and memory use matters.
- List comprehensions are useful when you need the complete result in memory.
- Performance depends on the task; choose clear code and measure before optimizing.
