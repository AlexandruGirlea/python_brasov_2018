### Python keeps loop variables local to list, dictionary and set comprehensions, and generator expressions.

```python
# Examples of local loop variables

a = 10

exp_1 = [a for a in range(4)]
print(exp_1)
# a is still 10.
a
```

```python
exp_2 = (a for a in range(5))
print(exp_2)
print(exp_2.__next__())
print(exp_2.__next__())
print(exp_2.__next__())
a
```

```python
exp_3 = dict((a, a**2) for a in range(5))
print(exp_3)
a
```

```python
exp_4 = set(a for a in [1,2,3,4,1,2,3])
print(exp_4)
a
```

```python
# A regular for loop changes the variable a.
for a in range(4):
    a += 1

# a has now changed.
a
```

```python
# Delete a and see what happens.
del(a)
```

```python
exp_1 = [a for a in range(4)]
print(exp_1)
# a no longer exists as a global name; the comprehension uses its own
# local loop variable.
a
```

```python
# Return a dictionary representing the current global symbol table
globals()
```

```python
# Use a generator when it fits the task.

# Generate unique lottery-number combinations.
from itertools import combinations

lottery_numbers = []
for a in range(1,50):
    lottery_numbers.append(a)

a = combinations(lottery_numbers, 6)

# Retrieve one combination immediately without building the entire list.
a.__next__()
```
