```python
# definim o functie care primeste 2 variabile ca argumente
def adding(a,b):
    c = a + b
    return c
```

```python
# adunam doua valori
adding(1,2)
```

```python
# ce se intampla daca nu oferim functiei parametrii necesari
adding()
```

Dupa cum putem vedea mai sus functia ne intoare o eroare, pentru ca lipsesc cei 2 parametrii neceasi, "a" si "b"
