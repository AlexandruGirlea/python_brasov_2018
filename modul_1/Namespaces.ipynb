## Namespace-urile sunt un mod de a mapa numele unui obiect catre obiectul stocat in memoria calculatorului.

Namespace-urile este acea denumire pe care o folosim pentru a extrage informatii din memoria calculatorului

```python
a = 1 # integer / numar intreg
```

```python
# folosind functia "id" putem afla locatia din memoria calculatorului care stocheaza valuarea variabilei "a"
# denumirea "a" este un "label" pentru valoarea 1
id(a)
```

```python
id(1)
```

Dupa cum putem vedea id(a) is id(1) ne dau exact aceeasi locatie din memoria calculatorului.

```python
# verificam locatia unei valori care nu are label
id(5)
```

```python
# mai adaugam un "label" b valorii initiale a variabilei "a", adica acuma si b = 1
# inseamna ca si id(b) trebuie sa fie egal cu id(1)
b = a
id(b)
```
