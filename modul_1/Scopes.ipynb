Un "Scope" este o regiune a programului Python unde un "namespace" poate fi accesat.

Scope tipurile:
- local;
- enclosing;
- global;
- built-in.

```python
# local scope

# a este disponibil numai in functia curenta
def local_scope():
    a = 123
    print(a)

# numele functiei urmat de paranteze activeaza functia si printeaza valuarea lui a, adica 123
local_scope()
```

```python
# daca incercam sa vedem ce valuare are namespace-ul "a" o sa primim o eroare pentru ca "a",
# este disponibil doar local in functia definita mai sus, si nu si in restul programului.
print(a)
```

```python
# enclosing scope

# valuarea lui "a" este definita in prima functie, iar a doua functie preia valuarea lui "a" si o foloseste
# pentru a printa valuarea lui "a"
def enclosing_scope():
    a = 100

    def new_local_scope():
        print('printing from local scope: ', a)
    
    new_local_scope()


enclosing_scope()
```

```python
# global scope

# "a" este disponibil de acum inainte in tot programul, pentru ca este scris liber(nu este in nici o functie)
a = 321
a
```

```python
# built-in scope

# sunt acele functii pe care python le are deja integrate, precum functia "len",
# care numara cate valori contine lista noastra "a"
a = [1,2,3,4]
len(a)
```
