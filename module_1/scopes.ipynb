A scope is a region of a Python program in which a namespace can be accessed.

Scope types:
- local;
- enclosing;
- global;
- built-in.

```python
# Local scope

# a is available only inside this function.
def local_scope():
    a = 123
    print(a)

# Calling the function prints the value of a: 123.
local_scope()
```

```python
# Accessing a here raises an error because it is local to the function above.
# That local name is unavailable in the rest of the program.
print(a)
```

```python
# Enclosing scope

# a is defined in the outer function. The inner function can access it
# and print its value.
def enclosing_scope():
    a = 100

    def new_local_scope():
        print('printing from local scope: ', a)
    
    new_local_scope()


enclosing_scope()
```

```python
# Global scope

# a is defined outside any function, so it is available at module scope.
a = 321
a
```

```python
# Built-in scope

# Python provides built-in functions such as len(),
# which counts the items in our list a.
a = [1,2,3,4]
len(a)
```
