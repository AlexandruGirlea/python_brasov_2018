# Data types

## Numbers

Numbers can also be complex, for example `a = 3.14j`.

```python
# Integer
a = 1
type(a)
```

```python
# Floating-point number
b = 2.3
type(b)
```

```python
# String: a sequence of characters
c = 'I am learning programming'
type(c)
```

```python
# Boolean Type
d = True
e = False
print(type(d))
print(type(e))
```

```python
# List
f = [1,2,3]
type(f)
```

```python
# tuple
g = (1,2,3)
type(g)
```

```python
# Dictionary: key-value pairs
h = {'a':1}
type(h)
```
