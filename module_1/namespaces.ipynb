## Namespaces map names to objects.

A namespace lets us refer to objects using names. Multiple names can refer to the same object.

```python
a = 1 # Integer
```

```python
# The id() function returns the identity of the object referred to by a.
# The name a is a label for the value 1.
id(a)
```

```python
id(1)
```

`id(a)` and `id(1)` return the same identity in this example.

```python
# Inspect the identity of a value without assigning it a name.
id(5)
```

```python
# Assign another name, b, to the object referred to by a.
# Therefore, id(b) is also equal to id(1) in this example.
b = a
id(b)
```
