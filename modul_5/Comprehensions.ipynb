## comprehensions: list , dict , and set .

````python
squares = []

for n in range(5):
    squares.append(n ** 2)
    
squares
````

````python
# sau putem sa o scriem cu lambda folosind functia map care aplica formula pt fiecare element in range(5)
squares_2 = map(lambda n: n**2, range(5))

list(squares_2)
````

````python
# comprehension list
# The list comprehension reads much better

[n ** 2 for n in range(5)]
````

## Definitie: 
```
A list comprehension consists of brackets containing an expression followed by
a for clause, then zero or more for or if clauses. The result will be a new list
resulting from evaluating the expression in the context of the for and if clauses
which follow it.
```

````python
# Nested comprehensions
items = '1234'
pairs = []
for a in range(len(items)):
    for b in range(a, len(items)):
        pairs.append((items[a], items[b]))

pairs
````

````python
pairs = [(items[a], items[b]) for a in range(len(items)) for b in range(a, len(items))]
pairs
````

# dict comprehensions

````python
from string import ascii_lowercase

lettermap = dict((c, k) for k, c in enumerate(ascii_lowercase, 1))
lettermap
````

````python
a = ['x','y', 'z']
b = dict()

for i, j in enumerate(a, 1):
    print(i, '|', j)
    b[i] = j
    
print(b)
````

````python
c = dict((i, j) for i, j in enumerate(a, 1))
c
````

# set comprehensions

````python
# set este o colectie de elemente neordonate si unice
word = 'servus'
letters_1 = set(c for c in word)
letters_1
````

````python
# expresia de mai sus este == cu expresia de mai jos
letters_2 = {i for i in word}
letters_2
````
