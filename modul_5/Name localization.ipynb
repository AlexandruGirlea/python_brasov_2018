### Python localizeaza variabilele din loop-uri in toate cele 4 forme | comprehensions: list , dict , set si generatori

```python
# exemple de localizare a variabilelor

a = 10

exp_1 = [a for a in range(4)]
print(exp_1)
# a este tot 10
a
```

```python
exp_2 = (a for a in range(5))
print(exp_2)
print(exp_2.__next__())
print(exp_2.__next__())
print(exp_2.__next__())
a
```

```python
exp_3 = dict((a, a**2) for a in range(5))
print(exp_3)
a
```

```python
exp_4 = set(a for a in [1,2,3,4,1,2,3])
print(exp_4)
a
```

```python
# for loop modifica variabila "a"
for a in range(4):
    a += 1

# variabila "a" a fost modificata
a
```

```python
# stergem variabila a sa vedem ce se intampla
del(a)
```

```python
exp_1 = [a for a in range(4)]
print(exp_1)
# variabila a nu mai exista ca si variabila globala, este doar o variabila temporara folosita local in declararea
# unui comprehension
a
```

```python
# concluzie folositi generatorii daca puteti

# calculam numarul de combinatii unice pentru nr la lot
from itertools import combinations

loto_numbers = []
for a in range(1,50):
    loto_numbers.append(a)

a = combinations(loto_numbers, 6)

# este instant
a.__next__()
```
