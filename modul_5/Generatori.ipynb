```python
# generatorii nu strang rezultatele unei iteratii si returneaza totul la sfarsit
# generatorii pot returna o singura valuare, se pot opri si pot continua prin returnarea unei noi valori
```

```python
 # classic function approach
def get_squares(n):
    return [x ** 2 for x in range(n)]
print(get_squares(3))
```

```python
def get_squares_gen(n): # generator approach
    for x in range(n):
        yield x ** 2 # we yield, we don't return
        
a = get_squares_gen(3)

print(next(a))
print(next(a))
# a generator will instead raise a StopIteration exception
```

```python
# generatorii mai au 3 metode: send , throw , si close

# send ne permite sa trimitem o valuare inapoi la generator
# throw ne permite sa ridicam o exceptie in generator
# close ne permite sa inchidem generatorul
```

```python
stop = True
def count(n=0):
    while stop:
        yield n
        n += 1
a = count()
print(a.__next__())
print(a.__next__())
print(a.__next__())
stop = False
try:
    count().__next__()
except StopIteration:
    print('exception raised StopIteration')
```

```python
# folosirea send()
def count(start=0):
    n = start
    while True:
        result = yield n
        print(type(result), result)
        if result == 'quit':
            break
        n += 1
        
c = count()
print(next(c))
print(c.send('ioi'))
print(next(c))
try:
    print(c.send('quit'))
except StopIteration:
    print('exception raised StopIteration')
```

## The yield from expression

```python
# we yield one at a time
def count(start, end):
    for n in range(start, end):
        yield n + 1

for n in count(2, 5):
    print(n)
```

```python
# yield from g is == for i in j: yield i
def count(start, end):
    yield from (n+1 for n in range(start, end))

for n in count(2, 5):
    print(n)
```
